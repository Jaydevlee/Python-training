```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score # 모델 평가 라이브러리

# xlsx 불러오기

df = pd.read_excel('./data/사과배복숭아-학습.xlsx')
# print(df)

# 결측지 값의 유무 확인
# print(df.isna().sum()) #isna() -> 결측치를 확인하는 함수

# 모든 결측치 값을0을로 대체
df.fillna(0, inplace=True) 
# print(df.isna().sum())

# fruit 칼럼의 항목값별 개수
# print(df['fruit'].value_counts())

# 과일이름을 숫자로 변경
# df.replace({'fruit':{'사과':0, '배':1, '복숭아':2}}, inplace=True) 최근에 비권장됨
df['fruit'] = df['fruit'].map({'사과':0, '배':1, '복숭아':2})
# print(df)

# print(df.groupby('fruit').mean())

# 그래프로 살펴보기 -> 인공지능이 학습하기 적합한 데이터를 파악하기 위함
# sns.scatterplot(data=df, x='weight', y='height', hue='fruit')

# 당도와 산도로 산점도
# sns.scatterplot(data=df, x='sweet', y='sour', hue='fruit') # 인공지능 학습으로 사용하기 적합하지 않음

# 모델 생성 및 학습
# 학습용 데이터
#                행, 열
x_train = df.iloc[:, 1:] # 학습 데이터 모든 행을 활용하고 fruit칼럼만 제외하고 
y_train = df.iloc[:, 0] # 정답 데이터, 모든행의 fruit열만 

#결정트리 모델을 사용해서 머신러닝 학습 모델 생성
# 학습 모델 객체 생성(알고리즘 생성)
model = DecisionTreeClassifier()

model.fit(x_train, y_train)
# 분류가 정상적으로 진행되었는지 확인
new_fruit = pd.DataFrame([[270.0, 75.0, 85.0, 37.0, 13.5, 0.22, 55.0]], columns=x_train.columns)

prd = model.predict(new_fruit)
print(prd)

# 모델 평가
# 테스트용 데이터
newdf = pd.read_excel('./data/사과배복숭아-테스트.xlsx')
newdf.fillna(0, inplace=True)
newdf['fruit'] = newdf['fruit'].map({'사과':0, '배':1, '복숭아':2})
print(newdf)

x_test = newdf.iloc[:, 1:] 
y_test = newdf.iloc[:, 0] 

tr_prd = model.predict(x_test)
print(tr_prd)

# 평가 하기
# 예측과 테스트 데이터의 정답을 비교해서 정확도 계산
print("정확도 = ", accuracy_score(y_test, tr_prd)) #accurancy_score: 정확도 계산 함수

for i in range(len(y_test)):
    print("예측 : ", tr_prd[i], "정답 : ", y_test.iloc[i])
```
